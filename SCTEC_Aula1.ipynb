📊 Resumo: Estatística Descritiva com NumPy
Este notebook demonstra a aplicação prática de métricas estatísticas utilizando a biblioteca NumPy. O objetivo é extrair insights precisos de conjuntos de dados, evitando interpretações equivocadas causadas por valores extremos.

🧰 Principais Funções
np.mean(dados): Calcula a média aritmética. Útil para cenários de somatório e projeções onde todos os valores possuem peso similar.

np.median(dados): Calcula a mediana. Essencial para encontrar o "valor típico" (o centro real), sendo imune a distorções causadas por outliers (valores muito altos ou baixos).

np.std(dados): Calcula o desvio padrão. Indica a volatilidade dos dados; quanto maior, mais dispersos estão os valores em relação à média.

np.min(dados) | np.max(dados): Identifica os limites inferior e superior do dataset.

💡 A "Regra de Ouro" para sua análise
Média vs. Mediana: Sempre compare ambas. Se a média estiver significativamente distante da mediana, isso é um indicativo claro de assimetria nos seus dados (presença de valores discrepantes).

Use Média quando quiser saber o total (ex: o faturamento total da empresa).

Use Mediana quando quiser saber o típico (ex: quanto o cliente comum realmente gasta).

🚀 Dica de boas práticas
Sempre que estiver analisando dados, verifique a distância entre a média e a mediana. Uma diferença muito grande entre elas é o seu primeiro sinal de alerta para investigar possíveis erros de entrada de dados ou comportamentos anômalos no seu sistema.
